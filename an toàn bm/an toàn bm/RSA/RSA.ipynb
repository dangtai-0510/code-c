```python
from math import gcd

# Kiem tra so nguyen to
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


# Ma hoa RSA
def mahoa(M, e, n):
    return pow(M, e, n)


# Giai ma RSA
def giaima(C, d, n):
    return pow(C, d, n)


# Nhap khoa
p = int(input("Nhap p: "))
q = int(input("Nhap q: "))

if not is_prime(p) or not is_prime(q) or p == q:
    print("p va q phai la hai so nguyen to khac nhau")

else:
    n = p * q
    phi = (p - 1) * (q - 1)

    print("n =", n)
    print("phi =", phi)

    e = int(input("Nhap e: "))
    d = int(input("Nhap d: "))

    if not (1 < e < phi) or gcd(e, phi) != 1:
        print("e khong hop le")

    elif not (1 <= d < phi) or (e * d) % phi != 1:
        print("d khong hop le")

    elif n <= 255:
        print("n phai lon hon 255 de ma hoa du lieu UTF-8")

    else:
        print("Public Key:", (e, n))
        print("Private Key:", (d, n))

        # Doc plaintext tu file
        with open("plaintext.txt", "r", encoding="utf-8") as f:
            text = f.read()

        print("Ban ro:", text)

        # Chuyen chuoi thanh cac byte
        data = text.encode("utf-8")

        # Ma hoa tung byte
        C = [mahoa(M, e, n) for M in data]

        # Luu ciphertext
        with open("ciphertext.txt", "w", encoding="utf-8") as f:
            f.write(" ".join(map(str, C)))

        print("Ban ma:", C)

        # Doc ciphertext tu file
        with open("ciphertext.txt", "r", encoding="utf-8") as f:
            cipher = list(map(int, f.read().split()))

        # Giai ma tung byte
        decoded = [giaima(x, d, n) for x in cipher]

        # Chuyen lai thanh chuoi
        result = bytes(decoded).decode("utf-8")

        # Luu plaintext sau giai ma
        with open("decrypted.txt", "w", encoding="utf-8") as f:
            f.write(result)

        print("Giai ma:", result)
        print("Da luu ket qua vao ciphertext.txt va decrypted.txt")
```
