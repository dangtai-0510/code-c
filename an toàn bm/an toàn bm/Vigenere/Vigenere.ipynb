```python

def mahoa(text, key):
    result = ""
    j = 0
    key = key.upper()

    for x in text:
        if 'A' <= x <= 'Z':
            k = ord(key[j % len(key)]) - ord('A')
            result += chr((ord(x) - ord('A') + k) % 26 + ord('A'))
            j += 1

        elif 'a' <= x <= 'z':
            k = ord(key[j % len(key)]) - ord('A')
            result += chr((ord(x) - ord('a') + k) % 26 + ord('a'))
            j += 1

        else:
            result += x

    return result


def giaima(text, key):
    result = ""
    j = 0
    key = key.upper()

    for x in text:
        if 'A' <= x <= 'Z':
            k = ord(key[j % len(key)]) - ord('A')
            result += chr((ord(x) - ord('A') - k) % 26 + ord('A'))
            j += 1

        elif 'a' <= x <= 'z':
            k = ord(key[j % len(key)]) - ord('A')
            result += chr((ord(x) - ord('a') - k) % 26 + ord('a'))
            j += 1

        else:
            result += x

    return result


print("1. Ma hoa")
print("2. Giai ma")
chon = int(input("Chon: "))
key = input("Nhap khoa: ").strip()

if not key or not key.isalpha() or not key.isascii():
    print("Khoa phai chi gom chu cai A-Z")
else:
    if chon == 1:
        with open("plaintext.txt", "r", encoding="utf-8") as f:
            text = f.read()

        result = mahoa(text, key)

        with open("ciphertext.txt", "w", encoding="utf-8") as f:
            f.write(result)

        print("Da ma hoa vao ciphertext.txt")

    elif chon == 2:
        with open("ciphertext.txt", "r", encoding="utf-8") as f:
            text = f.read()

        result = giaima(text, key)

        with open("plaintext_giaima.txt", "w", encoding="utf-8") as f:
            f.write(result)

        print("Da giai ma vao plaintext_giaima.txt")

    else:
        print("Lua chon khong hop le")
```
