```python
from math import gcd

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def ky_so(M, d, n):
    return pow(M, d, n)


def xac_minh(M, S, e, n):
    return pow(S, e, n) == M


p = int(input("Nhap p: "))
q = int(input("Nhap q: "))

if not is_prime(p) or not is_prime(q) or p == q:
    print("p va q phai la hai so nguyen to khac nhau")
else:
    n = p * q
    phi = (p - 1) * (q - 1)

    print("n =", n)
    print("phi =", phi)

    e = int(input("Nhap e: "))
    d = int(input("Nhap d: "))

    if not (1 < e < phi) or gcd(e, phi) != 1:
        print("e khong hop le")
    elif not (1 <= d < phi) or (e * d) % phi != 1:
        print("d khong hop le")
    else:
        print("Public Key:", (e, n))
        print("Private Key:", (d, n))

        M = int(input("Nhap thong diep M: "))

        if 0 <= M < n:
            S = ky_so(M, d, n)

            print("Chu ky so S =", S)

            with open("signature.txt", "w", encoding="utf-8") as f:
                f.write(str(S))

            with open("signature.txt", "r", encoding="utf-8") as f:
                S_doc = int(f.read())

            if xac_minh(M, S_doc, e, n):
                print("Chu ky hop le")
            else:
                print("Chu ky khong hop le")

            # Kiem tra khi thong diep bi thay doi
            M_sua = (M + 1) % n

            if xac_minh(M_sua, S_doc, e, n):
                print("Thong diep sua: Chu ky hop le")
            else:
                print("Thong diep sua: Chu ky khong hop le")
        else:
            print("M phai nam trong khoang 0 den", n - 1)
```
