```python
import hashlib

# Ham bam MD5
def md5(text):
    return hashlib.md5(text.encode("utf-8")).hexdigest()


# Doc ban ro tu file
with open("plaintext.txt", "r", encoding="utf-8") as f:
    text = f.read()

print("Ban ro:", text)

# Thuc hien bam MD5
H = md5(text)

print("Ma bam MD5:", H)

# Luu ket qua vao file
with open("hash_md5.txt", "w", encoding="utf-8") as f:
    f.write(H)

print("Da luu ket qua vao hash_md5.txt")
```
