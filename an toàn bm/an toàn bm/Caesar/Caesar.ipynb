```python

def mahoa(text, k):
    result = ""

    for x in text:
        if 'A' <= x <= 'Z':
            result += chr((ord(x) - ord('A') + k) % 26 + ord('A'))
        elif 'a' <= x <= 'z':
            result += chr((ord(x) - ord('a') + k) % 26 + ord('a'))
        elif '0' <= x <= '9':
            result += chr((ord(x) - ord('0') + k) % 10 + ord('0'))
        else:
            result += x

    return result


def giaima(text, k):
    return mahoa(text, -k)


print("1. Ma hoa")
print("2. Giai ma")
chon = int(input("Chon: "))
k = int(input("Nhap khoa k: "))

if chon == 1:
    with open("plaintext.txt", "r", encoding="utf-8") as f:
        text = f.read()

    result = mahoa(text, k)

    with open("ciphertext.txt", "w", encoding="utf-8") as f:
        f.write(result)

    print("Da ma hoa vao ciphertext.txt")

elif chon == 2:
    with open("ciphertext.txt", "r", encoding="utf-8") as f:
        text = f.read()

    result = giaima(text, k)

    with open("plaintext_giaima.txt", "w", encoding="utf-8") as f:
        f.write(result)

    print("Da giai ma vao plaintext_giaima.txt")

else:
    print("Lua chon khong hop le")
```
