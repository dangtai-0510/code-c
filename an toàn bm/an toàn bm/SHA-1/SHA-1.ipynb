```python
import hashlib

# Ham bam SHA-1
def sha1(text):
    return hashlib.sha1(text.encode("utf-8")).hexdigest()


# Doc du lieu tu file
with open("plaintext.txt", "r", encoding="utf-8") as f:
    text = f.read()

print("Ban ro:", text)

# Thuc hien bam SHA-1
H = sha1(text)

print("Ma bam SHA-1:", H)

# Luu ket qua vao file
with open("hash_sha1.txt", "w", encoding="utf-8") as f:
    f.write(H)

print("Da luu ket qua vao hash_sha1.txt")
```
